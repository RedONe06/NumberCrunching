````python
from __future__ import division # Garante divisão de números inteiros em double
import numpy as np # lib de computação científica
````

Convertendo de binário para decimal (LaTex):

\begin{align*}
x &= (1001, 101)_2 &\\
&= 1 \cdot 2^3 + 0 \cdot 2^2 + 0 \cdot 2^1 + 1 \cdot 2^0 + 1 \cdot 2^{-1} + 0 \cdot 2^{-2} + 1 \cdot 2^{-3} &\\
&= 8 + 0 + 0 + 1 + 0,5 + 0 + 0,125 &\\
&= 9,625 &
\end{align*}

````python
# Representando bases diferentes em python
print(0b1001) # bin -> dec
print(0o1357) # oct -> dec
print(0xbeba) # hex -> dec
````

Convertendo decimal em binário (Latex)

\begin{align*}
9,625 &= 9 + 0,625 \\
9 &= 4 \cdot 2 + 1 \\
4 &= 2 \cdot 2 + 0 \\
2 &= 1 \cdot 2 + 0 \\
1 &= 0 \cdot 2 + 1 \\
&= (1001)_2
\end{align*}

Em python:

````python
x = 9
d0 = x % 2
x = int(x/2)
print("d0 = %d, x = %d" % (d0, x))

d1 = x % 2
x = int(x/2)
print("d1 = %d, x = %d" % (d1, x))

d2 = x % 2
x = int(x/2)
print("d2 = %d, x = %d" % (d2, x))

d3 = x % 2
x = int(x/2)
print("d3 = %d, d4 = %d" % (d3, x))
````
